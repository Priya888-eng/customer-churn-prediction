```python
import pandas as pd

df = pd.read_csv('../data/cleaned/employees_cleaned.csv')
df.shape
```

```python
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})
df['Attrition'].value_counts()
```

```python
travel_map = {'Non-Travel': 0, 'Travel_Rarely': 1, 'Travel_Frequently': 2}
df['BusinessTravel'] = df['BusinessTravel'].map(travel_map)
df['BusinessTravel'].value_counts()
```

```python
df['OverTime'] = df['OverTime'].map({'Yes': 1, 'No': 0})
df['OverTime'].value_counts()
```

```python
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
df['Gender'].value_counts()
```

```python
df = pd.get_dummies(df, columns=['Department', 'EducationField', 'JobRole', 'MaritalStatus'], drop_first=True)
df.shape
```

```python
df.to_csv('../data/cleaned/employees_features.csv', index=False)
```
