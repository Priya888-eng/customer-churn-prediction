```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/cleaned/employees_cleaned.csv')
df.shape
```

```python
sns.kdeplot(data=df, x='Age', hue='Attrition', fill=True)
plt.title('Age Distribution by Attrition')
plt.show()
```

```python
df.groupby('Attrition')['Age'].mean()
```

```python
sns.countplot(data=df, x='JobSatisfaction', hue='Attrition')
plt.title('Job Satisfaction Level by Attrition')
plt.show()
```

```python
df.groupby('JobSatisfaction')['Attrition'].apply(lambda x: (x == 'Yes').mean() * 100).round(2)
```

```python
plt.figure(figsize=(14, 10))
sns.heatmap(df.select_dtypes(include='number').corr(), cmap='coolwarm', center=0)
plt.title('Correlation Heatmap of Numeric Features')
plt.show()
```
